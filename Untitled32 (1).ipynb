```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('RTA Dataset.csv')

print("Shape:", df.shape)
df.head()
```

```python
df.columns
df.info()
df.isnull().sum().sort_values(ascending=False)
```

```python
# Step 1: Standardize missing values
df = df.replace('na', np.nan)

# Step 2: Ensure target column is clean (safety step)
df = df.dropna(subset=['Accident_severity'])

# Step 3: Fill categorical columns with 'Unknown'
for col in df.columns:
    if df[col].dtype == 'object':
        df[col] = df[col].fillna('Unknown')

# Step 4: Fill numeric columns with median
df['Number_of_vehicles_involved'] = df['Number_of_vehicles_involved'].fillna(df['Number_of_vehicles_involved'].median())
df['Number_of_casualties'] = df['Number_of_casualties'].fillna(df['Number_of_casualties'].median())

# Step 5: Remove duplicates
df = df.drop_duplicates()
```

```python
# After cleaning
print("Duplicate rows:", df.duplicated().sum())

print("\nAccident Severity Distribution:")
print(df['Accident_severity'].value_counts())

print("\nWeather Conditions:")
print(df['Weather_conditions'].value_counts().head())
```

```python
# Convert Time column
df['Time'] = pd.to_datetime(df['Time'], format='%H:%M:%S', errors='coerce')

# Extract Hour
df['Hour'] = df['Time'].dt.hour

# Peak hour flag
df['Peak_Hour'] = df['Hour'].apply(lambda x: 1 if (7 <= x <= 9 or 16 <= x <= 18) else 0)
```

```python
df[['Time','Hour','Peak_Hour']].head()
```

```python
df['Day_of_week'].value_counts().plot(kind='bar')
plt.title("Accidents by Day of Week")
plt.xlabel("Day")
plt.ylabel("Count")
plt.show()
```

```python
df.groupby('Hour').size().plot(kind='line')
plt.title("Accidents by Hour")
plt.xlabel("Hour")
plt.ylabel("Count")
plt.show()
```

```python
df['Accident_severity'].value_counts().plot(kind='bar')
plt.title("Accident Severity Distribution")
plt.show()
```

```python
df['Cause_of_accident'].value_counts().head(10).plot(kind='bar')
plt.title("Top Causes of Accidents")
plt.show()
```

```python
df['Weather_conditions'].value_counts().head(10).plot(kind='bar')
plt.title("Weather Conditions Impact")
plt.show()
```

```python
pd.crosstab(df['Cause_of_accident'], df['Accident_severity']).head(10).plot(kind='bar', stacked=True)
plt.title("Severity vs Cause of Accident")
plt.show()
```

```python
pd.crosstab(df['Weather_conditions'], df['Accident_severity']).plot(kind='bar', stacked=True)
plt.title("Accident Severity by Weather Conditions")
plt.xlabel("Weather")
plt.ylabel("Count")
plt.show()
```

The graph illustrates the relationship between weather conditions and accident severity, showing that the majority of accidents occur under normal weather conditions, which is likely due to higher traffic volume rather than increased risk. Across all weather categories, slight injuries dominate, with comparatively fewer serious and fatal injuries. Although adverse conditions such as rain contribute to fewer overall accidents, they still show a noticeable proportion of more severe outcomes, suggesting that environmental factors can influence the severity of accidents even if they do not significantly increase their frequency.

```python
pd.crosstab(df['Day_of_week'], df['Accident_severity']).plot(kind='bar', stacked=True)
plt.title("Accident Severity by Day of Week")
plt.xlabel("Day")
plt.ylabel("Count")
plt.show()
```

The graph shows how accident severity varies across different days of the week, indicating that accident frequency is not evenly distributed throughout the week. Friday appears to have the highest number of accidents, followed by midweek days like Thursday and Wednesday, which may be associated with increased traffic volume and commuting activity. Across all days, slight injuries consistently dominate, with relatively fewer serious and fatal cases, suggesting that while the number of accidents fluctuates, the severity distribution remains similar. Overall, this pattern highlights the influence of temporal factors such as weekday traffic and driver behavior on accident occurrence.

```python
pd.crosstab(df['Peak_Hour'], df['Accident_severity']).plot(kind='bar', stacked=True)
plt.title("Accident Severity: Peak vs Non-Peak Hours")
plt.xlabel("Peak Hour (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.show()
```

The graph compares accident severity between peak and non-peak hours, showing that a higher number of accidents occur during non-peak hours compared to peak hours. This suggests that while traffic volume is higher during peak times, accidents are not necessarily more frequent, possibly due to slower driving speeds and increased caution. Across both categories, slight injuries dominate, with relatively fewer serious and fatal cases, indicating that the severity pattern remains consistent regardless of time. Overall, this highlights that factors beyond traffic density, such as driver behavior and road conditions, may play a significant role in accident occurrence.

```python
df.groupby('Accident_severity')['Number_of_casualties'].mean().plot(kind='bar')
plt.title("Average Casualties by Accident Severity")
plt.ylabel("Average Casualties")
plt.show()
```

The graph shows the average number of casualties associated with each level of accident severity, indicating a clear relationship between severity and impact. Fatal accidents have the highest average number of casualties, significantly higher than both serious and slight injury cases, which suggests that more severe accidents tend to involve greater harm and multiple individuals. In contrast, serious and slight injuries have relatively similar average casualty counts, indicating that while the injury level differs, the number of people involved may not vary as much. Overall, this highlights that accident severity is strongly associated with the extent of human impact, with more severe accidents leading to higher casualty outcomes.

```python
df.groupby('Accident_severity')['Number_of_vehicles_involved'].mean().plot(kind='bar')
plt.title("Average Vehicles Involved by Severity")
plt.ylabel("Average Vehicles")
plt.show()
```

The graph shows the average number of vehicles involved across different levels of accident severity, indicating that the number of vehicles involved does not increase with severity in a straightforward way. Slight injury accidents have the highest average number of vehicles involved, followed by serious and then fatal accidents, suggesting that more vehicles do not necessarily lead to more severe outcomes. This implies that accident severity may be influenced more by factors such as speed, impact type, and driver behavior rather than just the number of vehicles involved.

```python
top_causes = df['Cause_of_accident'].value_counts().head(5).index

pd.crosstab(df[df['Cause_of_accident'].isin(top_causes)]['Cause_of_accident'],
            df['Accident_severity']).plot(kind='bar', stacked=True)

plt.title("Top Causes vs Accident Severity")
plt.xlabel("Cause of Accident")
plt.ylabel("Count")
plt.show()
```

The graph illustrates the relationship between the top causes of accidents and their severity, highlighting that driver behavior is a major contributing factor across all severity levels. “No distancing” appears as the most frequent cause, followed by behaviors like changing lanes and driving carelessly, indicating that unsafe driving practices significantly contribute to accident occurrence. Across all causes, slight injuries dominate, but each cause still contributes to a noticeable number of serious and some fatal injuries. This suggests that while most accidents may be less severe, risky behaviors such as improper lane changes and lack of distancing can still lead to more serious outcomes, emphasizing the importance of driver awareness and adherence to safety practices.
